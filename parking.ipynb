총괄표제부에 주차가 있는 경우, 표제부에는 없는 경우가 있다.

둘을 비교해서 잘 골라야겠지.


그러면 주차는 총 3가지 방식으로 기록될 수 있다.

-   총괄표제부 총 주차 수
-   총괄표제부 옥내옥외기계자주×4
-   일반건축물대장/표제부 옥내옥외기계자주×4

총괄표제부가 없는 일반건축물대장/표제부는 한 가지 방식만 남으므로 그 방식으로 셀 수 있다.

총괄표제부가 있는 일반건축물대장/표제부는 3가지 방식의 비교가 필요하다.

3가지 방식 중 가장 많은 수를 진짜 총 주차 수로 사용한다.

그 외 옥내옥외기계자주×4 주차장 수는 따로 집계한다.


```python
df_site_complex = (
    df_pk_building.filter(pl.col("관리_상위_건축물대장_PK").is_not_null())
    .join(
        df_site.select(col_site),
        left_on="관리_상위_건축물대장_PK",
        right_on="관리_건축물대장_PK",
    )
    .join(df_dong.select(col_dong), "관리_건축물대장_PK", suffix="_dong")
    .with_columns(
        [
            # Calculate sum of indoor/outdoor mechanical and self-driven counts
            (
                pl.col("옥내_기계식_대수(대)")
                + pl.col("옥외_기계식_대수(대)")
                + pl.col("옥내_자주식_대수(대)")
                + pl.col("옥외_자주식_대수(대)")
            ).alias("옥내외_기계자주_대수_합계"),
            (
                pl.col("옥내_기계식_대수(대)_dong")
                + pl.col("옥외_기계식_대수(대)_dong")
                + pl.col("옥내_자주식_대수(대)_dong")
                + pl.col("옥외_자주식_대수(대)_dong")
            ).alias("옥내외_기계자주_대수_합계_dong"),
        ]
    )
    .group_by("관리_상위_건축물대장_PK")
    .agg(
        [
            pl.first("시군구_코드"),
            pl.first("주_용도_코드"),
            pl.first("주_용도_코드_명"),
            pl.col("주_용도_코드_dong").unique(),
            pl.max("세대_수(세대)"),
            pl.sum("세대_수(세대)_dong"),
            pl.max("총_주차_수"),
            pl.max("옥내외_기계자주_대수_합계"),
            pl.max("옥내_기계식_대수(대)"),
            pl.max("옥외_기계식_대수(대)"),
            pl.max("옥내_자주식_대수(대)"),
            pl.max("옥외_자주식_대수(대)"),
            pl.sum("옥내외_기계자주_대수_합계_dong"),
            pl.sum("옥내_기계식_대수(대)_dong"),
            pl.sum("옥외_기계식_대수(대)_dong"),
            pl.sum("옥내_자주식_대수(대)_dong"),
            pl.sum("옥외_자주식_대수(대)_dong"),
            pl.max("사용승인_일"),
            pl.max("사용승인_일_dong"),
        ]
    )
    .with_columns(
        [
            # Find the maximum value from the three methods of parking lot counts
            pl.max_horizontal(
                [
                    pl.col("총_주차_수"),
                    pl.col("옥내외_기계자주_대수_합계"),
                    pl.col("옥내외_기계자주_대수_합계_dong"),
                ]
            ).alias("최종_총_주차_수"),
        ]
    )
)

# .join(
#     df_site.select(col_site), "관리_건축물대장_PK"
# ).with_columns(
#     [
#         # Calculate sum of indoor/outdoor mechanical and self-driven counts
#         (
#             pl.col("옥내_기계식_대수(대)")
#             + pl.col("옥외_기계식_대수(대)")
#             + pl.col("옥내_자주식_대수(대)")
#             + pl.col("옥외_자주식_대수(대)")
#         ).alias("옥내외_기계자주_대수_합계")
#     ]
# )
df_site_complex.filter(pl.col("최종_총_주차_수") >= 10).head().collect()
```

```python
col_parking = [
    "시군구_코드",
    "주_용도_코드",
    "주_용도_코드_명",
    "최종_총_주차_수",
    # 기계식, 자주식, 옥내, 옥외 합계가 필요하면 추가 코딩 필요함. 그냥 갖다쓰면 안됨.
]
```

```python
# Select only the parking-related columns from both dataframes
# and Combine the two dataframes (union operation)
df_combined_parking = pl.concat(
    [df_site_complex.select(col_parking), df_building_single.select(col_parking)]
)

# Create a column to categorize the parking size into ranges (10대, 20대, 50대, 100대)
df_combined_parking = df_combined_parking.with_columns(
    [
        pl.when(pl.col("최종_총_주차_수") <= 10)
        .then(pl.lit("10대 이하"))  # Using pl.lit for literal strings
        .when(pl.col("최종_총_주차_수") <= 20)
        .then(pl.lit("11-20대"))
        .when(pl.col("최종_총_주차_수") <= 50)
        .then(pl.lit("21-50대"))
        .when(pl.col("최종_총_주차_수") <= 100)
        .then(pl.lit("51-100대"))
        .otherwise(pl.lit("100대 이상"))
        .alias("규모_구분")
    ]
)


# Group by 시군구, 용도, 규모 and aggregate the parking lot counts
df_grouped_parking = df_combined_parking.group_by(
    ["시군구_코드", "주_용도_코드", "주_용도_코드_명", "규모_구분"]
).agg(
    [
        pl.sum("최종_총_주차_수").alias("총_주차_대수_합계"),
    ]
)
sido_short = {
    "11": "서울",
    "21": "부산",
    "22": "대구",
    "23": "인천",
    "24": "광주",
    "25": "대전",
    "26": "부산",
    "27": "대구",
    "28": "인천",
    "29": "광주",
    "30": "대전",
    "31": "울산",
    "36": "세종",
    "41": "경기",
    "42": "강원",
    "43": "충북",
    "44": "충남",
    "45": "전북",
    "46": "전남",
    "47": "경북",
    "48": "경남",
    "49": "제주",
    "50": "제주",
    "51": "강원",
}
df_grouped_parking = df_grouped_parking.with_columns(
    [
        pl.col("시군구_코드")
        .str.slice(0, 2)
        .alias("시도_코드"),  # Extract the first two characters
        pl.col("시군구_코드")
        .str.slice(0, 2)
        .replace(sido_short)
        .alias("시도_약칭"),  # Then map 시도_약칭
    ]
)
# Reorder columns to place 시도_코드 and 시도_약칭 as the first columns
df_grouped_parking = df_grouped_parking.select(
    [
        "시도_코드",  # First column
        "시도_약칭",  # Second column
        "시군구_코드",
        "주_용도_코드",
        "주_용도_코드_명",
        "규모_구분",
        "총_주차_대수_합계",
    ]
)
df_grouped_parking.head().collect()
```

```python
df_grouped_parking.collect().to_pandas().to_csv(
    "output/parking/parking_general.csv", encoding="utf-8-sig", index=False
)
```

```python
# 시도 코드에 따른 집계
df_sido_grouped = (
    df_grouped_parking.group_by(["시도_코드", "시도_약칭"])
    .agg([pl.sum("총_주차_대수_합계").alias("주차_대수")])
    .sort("시도_코드")
)
df_sido_grouped.collect().to_pandas().to_csv(
    "output/parking/parking_by_sido.csv", encoding="utf-8-sig", index=False
)
df_sido_grouped.head().collect()
```

```python
# 주용도 코드에 따른 집계
df_usage_grouped = (
    df_grouped_parking.group_by(["주_용도_코드", "주_용도_코드_명"])
    .agg([pl.sum("총_주차_대수_합계").alias("주차_대수")])
    .sort("주_용도_코드")
)
df_usage_grouped.collect().to_pandas().to_csv(
    "output/parking/parking_by_usage.csv", encoding="utf-8-sig", index=False
)
df_usage_grouped.head().collect()
```
